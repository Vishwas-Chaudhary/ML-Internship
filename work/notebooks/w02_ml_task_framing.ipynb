## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
print("""
Target / proxy:

The observed outcome of interest is page CTR and engagement performance.
For the eventual ML model, the target should come from an observed outcome
in a later time window rather than from a hand-written threshold.

The current starter dataset provides measured CTR, clicks, impressions,
sessions, and engagement metrics that help define the unit and the problem.
A future-window outcome can be used as the model target when the time-based
warehouse data is available.

I will not define the target using a rule such as CTR < 0.10.
""")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
print("""
Why ML beats a fixed rule:

A fixed rule such as "review every page with CTR below a chosen
threshold" is too simple because CTR and engagement can depend on
multiple factors.

These factors can include position tier, content type, search demand,
content characteristics, impressions, and other page-level signals.

An ML scoring model can combine multiple signals and rank pages for
review instead of relying on one manually chosen threshold.

This approach is useful for decision support because it can consider
different page characteristics together rather than treating every
page the same.

The model should be evaluated on whether its rankings generalize to
new observations, not used as proof that one factor causes another.
""")
```
